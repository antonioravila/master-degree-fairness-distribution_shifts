```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
os.chdir(r'C:\master-degree-fairness-distribution_shifts')

df = pd.DataFrame()
columns = ['DT_NOTIFIC', 'EVOLUCAO', 'CS_RACA', 'CS_SEXO', 'NU_IDADE_N']

for path in Path('src\data\SRAG').iterdir():
    print(path)
    df_raw = pd.read_parquet(path, columns = columns)
    df = pd.concat([df, df_raw])

df.dropna(inplace=True)
df['EVOLUCAO'] = df['EVOLUCAO'].astype(int)
df.to_parquet(path='src\data\SRAG\srag.parquet')

df
```

##  Kernel Distribution Plots

```python
state = 'TX'
acs_type = 'Income'

target_folder = rf'src\charts\ACS{acs_type}\{state}'
if not os.path.exists(target_folder):
    os.makedirs(target_folder)

for col in df.columns:
    plt.figure()
    sns.kdeplot(
        data=df, x=col, hue='Year', fill=True, common_norm=False, alpha=0.5, linewidth=0
    )
    plt.title(f"'{col}' distribution per year ")

    file_name = rf'\{col}-distribution-per-year.png'

    plt.savefig(target_folder + file_name)
```

```python
target_folder = rf'src\charts\compas'
if not os.path.exists(target_folder):
    os.makedirs(target_folder)

columns = [
    'age',
    'priors_count',
    'days_b_screening_arrest',
    'c_charge_degree',
    'race',
    'is_recid',
    'screening_date',
]

for col in df[columns].columns:
    plt.figure()
    sns.kdeplot(
        data=df, x=col, hue='Year', fill=True, common_norm=False, alpha=0.5, linewidth=0
    )
    plt.title(f"'{col}' distribution per year ")

    file_name = rf'\{col}-distribution-per-year.png'

    plt.savefig(target_folder + file_name)
```
